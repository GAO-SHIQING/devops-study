## 为什么需要控制流？

上一课的程序是"直线"的：从上到下，逐行执行，没有分叉，没有回头。

但真实世界不是这样的：
- **如果**下雨，带伞；**否则**不用带
- **重复**做俯卧撑，直到做完 30 个
- 数 1 到 100，**跳过**所有带 7 的数字

控制流（control flow）就是让程序有能力做**决策**和**重复**。没有控制流，程序就是一条没有岔路也没有回头的死路。有了控制流，程序才有了真正的"生命力"。

> **类比**：没有控制流的程序像一张购物清单——从头念到尾就完了。有控制流的程序像一个交互式的菜谱——"如果面团太干就加点水，重复揉面直到表面光滑"——它根据情况做决策，并且能重复做某件事。

### 更深一层：为什么控制流如此重要？

想象你要开发一个**自动咖啡机**的程序。没有控制流，你只能写"加咖啡豆→加水→加热→出杯"——但万一是顾客点了冰咖啡呢？万一水箱没水了呢？万一咖啡豆用完了呢？没有控制流，程序无法应对这些"万一"。

控制流的本质是：**让程序能根据不同的输入和环境状态，做出不同的响应**。这是程序从"死"到"活"的关键一步。

### 三大控制流结构

| 类型 | 作用 | 日常类比 |
|------|------|----------|
| **顺序** | 逐行执行（上一课已学） | 按菜谱步骤依次操作 |
| **选择** (if) | 根据条件走不同分支 | 岔路口选一条路 |
| **循环** (for/while) | 重复执行某段代码 | 绕圈跑，直到跑够圈数 |

本课将深入讲解**选择**和**循环**的方方面面，并揭示 Python 在这些方面的独特设计哲学。

## 1. 条件判断：if / elif / else

`if` 语句是编程中最基本的决策工具。它的结构很简单：

```python
if 条件:
    做某事
elif 其他条件:
    做另一件事
else:
    以上都不满足时做这个
```

### 真实的执行过程（逐行跟踪）

想象 Python 看到一段 `if` 语句时，它的大脑里发生了什么：

```python
score = 85

if score >= 90:         # 第1步：判断 85 >= 90 ? → False，跳过这个分支
    print("优秀")
elif score >= 80:       # 第2步：判断 85 >= 80 ? → True！进入这个分支
    print("良好")        # 第3步：执行 print("良好")，输出"良好"
elif score >= 60:       # 第4步：注意！——前面的分支已经执行了，这个和后面的 else 都被跳过
    print("及格")
else:
    print("不及格")
```

**关键是**：一旦某一个分支的条件满足并执行了，Python 就会**跳出整个 if-elif-else 结构**，不会再判断后面的条件。这叫做"短路"——找到第一个满足的条件就停。

### 为什么不用多个 if？

如果用多个独立的 `if`：

```python
if score >= 90: print("优秀")
if score >= 80: print("良好")   # score=85 时，这个也会执行！
```

score=85 时，第一个 if 不满足，但第二个 if 满足，会输出"良好"。这没问题。但如果有 `>=60` 的 if，score=85 时它也会满足——你会同时看到"良好"和"及格"。通常这**不是你想要的**。`elif` 保证了互斥：多个条件中最多只有一个被执行。

> **另一个类比**：if-elif-else 就像多层电梯按钮。你按了"3楼"，电梯会在 3 楼停下，不会继续往上到 4 楼、5 楼。而多个独立的 if 就像你同时按了 3、4、5 楼——每一层它都会停。

### 常见陷阱

1. **条件顺序搞反**：如果把 `score >= 60` 放在 `score >= 80` 前面，score=85 时会先匹配"及格"——逻辑错误！**更严格的条件要放在更宽松的条件前面。**
2. **忘记冒号**：`if score >= 80` 后面忘了冒号 → `SyntaxError`
3. **用 `=` 代替 `==`**：`if score = 80` 是赋值，不是比较。Python 中会直接报错（和其他语言不同，这反而保护了你）

````python
# ================================
# 2.1 if / elif / else 基础
# ================================

# 定义一个变量 score，赋值为 85，表示考试成绩
# 后续的条件判断将基于这个值来决定输出什么等级
score = 85

# if 语句：判断 score 是否 >= 90
# 条件表达式 score >= 90 的计算结果是 False（因为 85 < 90）
# 因此这个 if 代码块被跳过，不会执行 print("优秀")
if score >= 90:
    print("优秀")
# elif = "else if" 的缩写——当前面的 if 条件不满足时，才来检查这个条件
# 判断 score >= 80 → 85 >= 80 成立，结果为 True
# 关键特性：一旦某个 elif 分支的条件满足并执行，Python 就会跳出整个 if-elif-else 结构
# 不会再判断后续的任何条件——这保证了分支的互斥性
elif score >= 80:
    # 执行到这里，输出"良好"
    print("良好")
elif score >= 60:
    # 虽然 85 >= 60 也成立，但由于上面的 elif 已经执行了
    # Python 根本不会检查这个条件——这是 elif 与多个独立 if 的核心区别
    print("及格")
else:
    # 当以上所有条件都不满足时才会执行这里（兜底分支）
    # 本例中 score=85，不会执行到这里
    print("不及格")
# 输出：良好
# 执行流程总结：85 >= 90? No → 85 >= 80? Yes → 输出"良好" → 跳出整个 if 结构

# ---------- 对比：多个独立 if ----------
print("\n--- 多个独立 if 的效果 ---")
s2 = 85  # 重新定义一个变量 s2，便于与上面的 elif 版本对照
if s2 >= 90:
    print("优秀")  # 条件不满足（85 < 90），不执行
if s2 >= 80:
    print("良好")    # 这个也会执行！因为每个 if 是独立的判断，互不影响
if s2 >= 60:
    print("及格")    # 这个也会执行！问题：score=85 时同时输出"良好"和"及格"
# 用独立 if 时 score=85 会同时输出"良好"和"及格"——通常不是我们想要的（分数分段应该是互斥的）
# elif 保证了多个条件中最多只有一个被执行，这才是分数分段的正确逻辑

# ---------- 三目运算符（简洁版 if-else） ----------
# 三目运算符（ternary operator）语法：值1 if 条件 else 值2
# 当条件为 True 时取"值1"，否则取"值2"
# 适合简单的二选一赋值场景，能让代码更紧凑
age = 20
status = "成年" if age >= 18 else "未成年"
# f-string（格式化字符串）：用花括号 {} 在字符串中嵌入变量或表达式
print(f"\n{age}岁: {status}")

# 上面的三目运算符等价于以下完整的 if-else 结构：
if age >= 18:
    status = "成年"
else:
    status = "未成年"
````

## 2. 缩进——Python 的"紧箍咒"

### 深入理解：为什么 Python 强制缩进？

几乎所有语言都用花括号 `{}` 表示代码块。Python 是极少数用**缩进**表示代码块的语言。

**为什么这么设计？** Python 之父 Guido van Rossum 认为：

1. **代码同时也是写给人类读的**——缩进本来就应该反映代码的逻辑结构。既然你写代码时**本来就会缩进**来让代码可读，那为什么不直接把缩进变成语法的一部分呢？
2. **强制缩进 = 强制好习惯**——所有 Python 代码看起来都整洁统一。你永远不会看到"花括号在同一行"vs"花括号在下一行"的圣战。
3. **减少语法噪音**——没有 `{}`、没有 `end`、没有 `begin...end`，更干净，更少输入。

> **一个有趣的实验**：找一段 C 或 Java 代码，手动把所有花括号删掉，光靠缩进还能看懂逻辑吗？大概率可以。那反过来：把缩进去掉，只留花括号呢？可读性会严重下降。Guido 的洞察就是：**既然缩进才是人类真正用来理解代码结构的线索，那就让它成为规则本身。**

其他语言："你要加花括号表示这是 if 里面的代码。哦对了，最好也缩进一下让人看得懂。"
Python："**缩进本身就是语法。** 你怎么写 = 程序怎么走。"

### 缩进规则

- 同一代码块中的所有行，缩进必须**完全一致**
- Python 官方推荐：**4 个空格**（不是 Tab，虽然 Tab 也可以，但千万别混用！）
- 混用 Tab 和空格会导致 `IndentationError`——极其常见的初学者错误

### 常见的缩进错误

```
if True:
print("缩进错误")       # IndentationError: expected an indented block

if True:
    print("对的")
  print("缩进不一致")    # IndentationError: unindent does not match
```

### Python 内部的缩进机制

当 Python 解析代码时，它在内存中维护一个**缩进栈**：

1. 看到 `if` 行末尾的冒号 `:` → 入栈一个缩进级别（预期下一行缩进更多）
2. 下一行缩进多了 4 个空格 → Python 知道这是代码块开始
3. 遇到缩进减少的行 → Python 知道代码块结束了
4. 如果缩进量无法匹配栈中任何级别 → `IndentationError`

这就是为什么 Python 的缩进不仅是"风格问题"，而是**语法规则**。解析器靠缩进量来判断代码块的开始和结束。

### 实际建议

- **配置编辑器**：让编辑器自动把 Tab 键转换为 4 个空格。几乎所有的 Python 开发者都这么做。
- **查看缩进**：在编辑器中开启"显示空白字符"功能，可以清楚看到是 Tab 还是空格。
- **一个常见的坑**：从网页上复制代码时，有时会复制到不可见的 Tab 字符，导致看似对齐但实际上缩进不一致。建议复制后重新手动敲入缩进。

> **记住**：在 Python 中，"对齐"不仅仅是美观问题——它是程序逻辑的一部分。看到缩进错了，你的程序就可能做着完全不一样的事情（或者直接崩溃）。

````python
# ================================
# 2.2 缩进演示
# ================================

# 正确的缩进：同一级别的代码必须对齐
x = 10
# if 条件判断：x > 5 为 True（因为 10 > 5），进入下面的代码块
if x > 5:
    # 缩进 4 个空格——Python 用缩进来定义"这个代码属于 if 语句块"
    # 如果这一行没有缩进，Python 会抛出 IndentationError，程序直接崩溃
    print("x 大于 5")       # 缩进 4 个空格
    # 嵌套 if：在 if 内部再做一次条件判断
    # 缩进 8 个空格 = 外层 if 的 4 空格 + 内层 if 的 4 空格
    if x > 8:
        # 只有 x > 5 且 x > 8 两个条件同时满足时才会执行
        # 这里 x=10，两个条件都成立，所以会执行
        print("x 也大于 8")  # 嵌套 if，缩进 8 个空格
    # 缩进 4 个空格——和第一个 print 对齐，说明它属于外层 if，不属于内层 if
    # 重要：无论内层 if 的条件是否满足，只要外层 if 成立，这一行都会执行
    print("这句话属于外层 if")  # 缩进 4 个空格，和第一个 print 对齐
# 没有缩进——与 if 平级，说明它不属于 if 代码块
# 不管 if 的条件是否成立，这一行都会执行（因为它在 if 结构之外）
print("这句话在 if 外面")      # 没有缩进，和 if 平级

# 下面这行取消注释后会报错：
#  if True:        # 注意 if 前面有一个空格！属于不一致的缩进
# print("缩进不对")  # IndentationError: unexpected indent
# Python 不允许在同一个文件中混用不一致的缩进

# ---------- 缩进决定了逻辑 ----------
# 思考：下面两段代码有什么区别？
n = 3
if n > 0:
    print("正数")        # n > 0 为 True（3 > 0），执行
    # 注意：这一行有 4 个空格的缩进，属于 if 代码块
    # 所以只有当 n > 0 时才会执行——和上面的 print 在同一个代码块中
    print("这个也属于 if")
# 这一行没有缩进，在 if 外面，无论如何都会执行
print("这个在 if 外面")

print("---")
# 对比：
if n > 0:
    print("正数")
# 以下两行都没有缩进，不属于 if 代码块
# 与上面代码的核心区别：上面代码中有两个 print 隶属于 if 内部
# 这里只有一个 print 隶属于 if 内部，另外两个在 if 外面
# 缩进的差异直接改变了程序的逻辑行为！
print("这个在 if 外面")
print("这个也在 if 外面")
````

## 3. 逻辑运算符：and / or / not

### 为什么需要逻辑运算符？

很多时候一个条件不够用：
- "如果温度 > 30 **并且** 湿度 > 80，开除湿机"——两个条件同时满足
- "如果今天是周六 **或者** 是周日，不用上班"——两个条件之一满足
- "如果**不是**会员，加收服务费"——取反

生活中的逻辑判断几乎都是**复合条件**，不是单一条件。逻辑运算符就是用来"组合条件"的。

### 真值表

| A | B | A and B | A or B | not A |
|---|---|---------|--------|-------|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

> **记忆口诀**：`and` 找 False（有一个 False 就全 False），`or` 找 True（有一个 True 就全 True），`not` 是"反着来"。

### 短路求值——一个关键但容易被忽略的机制

Python 在计算 `and` 和 `or` 时，不会"两边都算完再结合"，而是**从左到右，能停就停**：

**`A and B`**：如果 A 是 False，整个表达式一定是 False，B **根本不会被执行**。
- 例子：`user and user.login()` —— 如果 user 是 None，右侧的 `user.login()` 永远不会调用，从而避免了 `AttributeError`。这是 Python 中非常常见的"安全访问"模式。

**`A or B`**：如果 A 是 True，整个表达式一定是 True，B **根本不会被执行**。
- 例子：`name or "默认用户"` —— 如果 name 是空字符串（falsy），就返回"默认用户"。

> **类比**：短路求值就像你打电话问朋友要不要一起吃饭。对于 `and`：朋友说"我不去"——你不需要问第二个朋友了，结果已经定了。对于 `or`：朋友说"我去"——同样不用再问别人了。

### 短路的好处
1. **性能**：有时第二个条件的计算很昂贵（比如调用数据库、访问网络），能省则省
2. **安全**：可以用 `if obj and obj.attr` 安全地访问可能不存在的属性
3. **简洁**：可以用 `x = a or b` 实现"如果 a 有效就用 a，否则用 b"的默认值模式

### 常见陷阱

**运算符优先级问题** —— `not` > `and` > `or`：

```python
# 你可能以为它在检查"年龄大于18且是会员，或者年龄大于60"
if age > 18 and is_member or age > 60:
    ...

# 但 Python 实际理解为：
if (age > 18 and is_member) or age > 60:
    ...

# 如果实际想要的是"年龄大于18，且（是会员或年龄大于60）"：
if age > 18 and (is_member or age > 60):
    ...
```

**最佳实践**：当组合多个逻辑运算符时，**用括号明确优先级**。这不只是为了正确性，更是为了让读你代码的人（包括未来的你自己）能一眼看懂你的意图。

````python
# ================================
# 2.3 逻辑运算符
# ================================

# ---------- and：两个都满足 ----------
# 定义两个变量：温度和湿度
temp = 35
humidity = 70
# and 运算符：左右两个条件都必须为 True，整个表达式才为 True
# 执行流程：先算 temp > 30（True），再算 humidity > 60（True）
# True and True → True，进入 if 代码块
if temp > 30 and humidity > 60:
    print("又热又闷，开空调！")

# ---------- or：至少一个满足 ----------
day = "周六"
# or 运算符：左右两个条件中只要有一个为 True，整个表达式就为 True
# 执行流程：先算 day == "周六"（True），已经确定了 or 的结果为 True
# 右侧的 day == "周日" 不会再计算了——这叫"短路求值"
if day == "周六" or day == "周日":
    print("今天休息，不上班！")

# ---------- not：取反 ----------
is_raining = False
# not 运算符：把 True 变成 False，把 False 变成 True
# not False → True，所以 if 条件成立
if not is_raining:
    print("不下雨，出门散步！")

# ---------- 短路求值演示 ----------
# 定义两个函数，它们会打印被调用的信息并返回布尔值
# 用函数来演示是因为函数调用是"可见的"——你能看到哪个函数被执行了
def check_left():
    print(">>> check_left 被调用了")
    return False  # 返回 False，表示"左侧条件不满足"

def check_right():
    print(">>> check_right 被调用了")
    return True   # 返回 True，表示"右侧条件满足"

print("=== and 短路演示 ===")
# and 运算：从左到右执行，左侧为 False 时，整个表达式一定是 False
# 因此 check_right() 根本不会被调用！
result = check_left() and check_right()
# 注意输出：只看到 check_left 被调用，没有 check_right 的输出！
# 这就是短路求值——能省则省，不计算不需要的部分

print("\nor 短路演示===")
# or 运算：从左到右执行，左侧为 False 时，不能确定 or 的结果
# 因为左侧 False 时结果取决于右侧，所以必须调用 check_right()
result = check_left() or check_right()
# check_left() 返回 False → 无法确定 or 的结果 → 继续调用 check_right()
# check_right() 返回 True → 最终结果为 True

# ---------- 日常使用模式 ----------
# 安全访问（利用 and 短路）
user = None
# 当 user 为 None 时，None 是 falsy 值，and 左侧为 False
# 因此右侧的 user.name 根本不会执行——避免了 AttributeError
# 这就是 Python 中常见的"安全导航"模式
if user and user.name:
    print("用户名:", user.name)
else:
    print("没有用户")

# 默认值模式（利用 or 短路）
input_name = ""
# or 运算符：左侧为 falsy 值（空字符串）时，返回右侧的值
# 左侧为 truthy 值时，返回左侧的值（不计算右侧）
# 这里 input_name 是空字符串（falsy），所以 name 被赋值为 "访客"
name = input_name or "访客"
print(f"你好，{name}")  # 输出：你好，访客（因为 input_name 是空字符串，falsy）
````

## 4. 真值（Truthiness）——"空"就是 False

### 为什么需要理解真值？

Python 中**任何值**都可以被当作布尔值来用。这意味着你不必写 `if len(list) > 0`，直接写 `if list` 就够了。

但这也带来了一个必须理解的问题：**什么值是"假的"？**

### Python 中所有的 Falsy 值

```python
if 值:   # 这个值会被自动转成 bool 判断
```

以下所有值都会被当成 **False**：

| 类型 | Falsy 值 | 为什么是 False |
|------|----------|----------------|
| bool | `False` | 就是假 |
| int | `0` | "没有"数量 |
| float | `0.0` | "没有"数量 |
| str | `""` (空字符串) | "没有"字符 |
| list | `[]` (空列表) | "没有"元素 |
| dict | `{}` (空字典) | "没有"键值对 |
| tuple | `()` (空元组) | "没有"元素 |
| set | `set()` | "没有"元素 |
| range | `range(0)` | "没有"数字 |
| None | `None` | "什么都没有" |

**其他所有值都是 Truthy**（包括负数和"False"这个字符串本身！）。

### 为什么这样设计？

这反映了 Python 的一个核心理念：**"空"就是 False**。不管是空的容器、零的数量、还是没有值，在逻辑上都对应"假"。这让代码更简洁：

```python
# 不推荐（但很多初学者这样写）
if len(items) > 0:
    print("有内容")

# 更 Pythonic
if items:
    print("有内容")
```

> **类比**：就像问"冰箱里有吃的吗？"对于空冰箱（空列表），答案是"没有"；对于有东西的冰箱，答案是"有"。Python 的真值规则就是这种直觉的自然延伸。

### 常见的坑

```python
if "False":   # 这是 True！！！因为非空字符串都是 True
if 0:         # 这是 False
if -1:        # 这是 True！！！负数也是非零
if "":        # 这是 False
if " ":       # 这是 True！！！空格不是空
```

**最坑的一个**：`bool("False")` 是 `True`——因为 `"False"` 是一个非空字符串！如果你需要判断字符串内容是不是 `"False"`，要用 `s == "False"`，不能依赖真值转换。

### 利用真值简化代码的常见模式

```python
# 模式1：检查列表是否为空
if items:           # ✅ Pythonic
if len(items) > 0:  # ❌ 啰嗦，不推荐

# 模式2：检查变量是否已赋值
if name:            # name 非空字符串时才执行（空字符串会被跳过）
    print(f"你好，{name}")

# 模式3：提供默认值
name = input_name or "访客"  # 如果 input_name 是空/falsy，用"访客"

# 模式4：递归终止条件
def countdown(n):
    if n:            # n=0 时为 False，递归终止
        print(n)
        countdown(n - 1)
```

> **记忆技巧**：想一下"如果某个东西是零、空、或者没有——它就代表 False"。否则就是 True。这个规则在 Python 中非常一致，几乎不需要死记硬背。

````python
# ================================
# 2.4 真值（Truthiness）
# ================================

# 所有在 if 条件中会被当作 False 的值（Falsy 值）
# 这是 Python 中一个非常重要的概念——记住它们可以避免很多 bug
falsy_values = [False, 0, 0.0, "", [], {}, (), None, set()]

print("=== Falsy 值列表 ===")
# 遍历 falsy 值列表，逐个验证它们的布尔值
# val!r 表示用 repr() 格式显示值，加上 :8 表示左对齐占 8 个字符宽度
for val in falsy_values:
    print(f"bool({val!r:8}) = {bool(val)}")  # 全部输出 False

print()

# ---------- 容易被误解的情况 ----------
print("=== 容易搞错的例子 ===")
# 以下是一些初学者经常判断错误的例子
print(f"bool(-1)      = {bool(-1)}")        # True！非零整数（即使是负数）都是 True
print(f"bool(0.001)   = {bool(0.001)}")      # True！非零浮点数也是 True
print(f"bool('False') = {bool('False')}")    # True！非空字符串都是 True，内容无关
print(f"bool(' ')     = {bool(' ')}")        # True！空格字符不是空字符串，所以是 True

# ---------- 实际应用：简化代码 ----------
def process_items(items):
    # Pythonic 写法：直接判断容器是否为空
    # if not items 等价于 if len(items) == 0，但更简洁
    # 空列表 [] 是 falsy → not [] → True → 进入 if
    if not items:
        print("没有需要处理的数据")
        return  # 提前退出函数，不再执行后面的代码
    print(f"处理 {len(items)} 项数据...")

process_items([])      # 传入空列表，输出"没有需要处理的数据"
process_items([1, 2])  # 传入非空列表，输出"处理 2 项数据..."

# ---------- 经典坑 ----------
name = "False"
# 注意：字符串 "False" 不是空字符串，所以它是 truthy 值
# 初学者常犯的错误：认为字符串 "False" 对应布尔值 False
# 实际上，只有空字符串 "" 才是 falsy
if name:
    print("这行会执行！因为'False'这个字符串不是空的")
# 期望：name 是 "False" → 以为不会执行 → 但实际上会执行！
````

## 5. for 循环——"给我一个序列，我逐个处理"

### 为什么需要循环？

写程序时，你经常需要**对一批数据做同样的事**：
- 对列表中的每个数字求平方
- 处理文件中的每一行
- 读取数据库中的每条记录

没有循环，你要写重复的代码一百次。有循环，你写一次，告诉电脑"重复执行直到处理完"。

> **类比**：for 循环就像流水线上的工人。传送带送来一个零件，你处理一个；再来一个，再处理一个……直到所有零件处理完。你不需要知道一共有多少个零件——传送带停了，你就下班。

### for 循环的本质

在 Python 中，`for` 循环的语法是：

```python
for 变量 in 可迭代对象:
    处理这个变量
```

**执行过程**（Python 的内心独白）：
1. 拿到可迭代对象（比如列表 `[10, 20, 30]`）
2. 问它："你内部有迭代器吗？给我一个"
3. 一次次问迭代器："下一个是什么？"
4. 迭代器给出下一个值 → 赋给变量 → 执行循环体
5. 迭代器说"没有了" → 循环结束

这个过程叫做**迭代协议**。`for` 循环在底层就是这么工作的——它不是魔术，只是一个"拿下一个，处理，再拿下一个"的重复过程。

### 可以迭代什么？

- 列表、元组、字符串、字典、集合、range、文件...
- **任何实现了迭代协议的对象**

字符串迭代时逐个给出字符，字典迭代时逐个给出键。

### 为什么叫"变量"？

`for x in list` 中的 `x` 就是**循环变量**。每次循环，Python 从序列中取出一个元素放进这个变量。这个名字你可以任意取，但好的命名让代码自解释：

```python
for student in students:   # 好——一眼看出每个元素是什么
for i in list:             # 差——i 通常表示索引，但这里拿到的是元素
```

### 常见陷阱

**1. 在循环中修改正在遍历的列表**

```python
numbers = [1, 2, 3, 4, 5]
for n in numbers:
    if n % 2 == 0:
        numbers.remove(n)  # 危险！修改了正在遍历的列表
# 结果可能出乎意料——因为 remove 会改变列表长度和元素位置
```

当你遍历列表时，Python 内部使用索引跟踪当前位置。如果在遍历过程中删除元素，后面的元素会往前移，导致某些元素被跳过。**如果需要过滤列表，建议创建一个新列表或使用列表推导式。**

**2. 循环变量在循环结束后仍然存在**

```python
for i in range(5):
    pass
print(i)  # 输出 4！i 在循环结束后仍然存在
```

这在 Python 中不会报错，但可能会导致后续代码意外使用到这个残留的变量值。如果之后有代码用了 `i`（比如另一个循环），不会报错但逻辑可能出错。Python 中 `for` 循环的变量没有自己的作用域——它和循环外部的代码在同一作用域中。

### for vs while：什么时候用哪个？

| 场景 | 推荐用 | 原因 |
|------|--------|------|
| 遍历列表/字符串等序列 | `for` | 直接、简洁 |
| 知道确切循环次数 | `for` + `range()` | 自动管理循环变量 |
| 不知道循环次数 | `while` | 靠条件控制 |
| 需要循环变量持续递增 | `for` + `range()` | 不容易忘记更新变量 |
| 需要复杂的退出条件 | `while` | 条件灵活 |

**简单的判断标准**：如果问题可以表述为"对**每个**元素做某事"，用 `for`；如果问题可以表述为"**一直**做某事，**直到**某个条件满足"，用 `while`。

````python
# ================================
# 2.5 for 循环基础
# ================================

# ---------- 遍历不同类型 ----------
# 遍历列表
fruits = ["苹果", "香蕉", "橘子"]
# for 循环：从 fruits 列表中逐个取出元素，赋值给循环变量 fruit
# 每轮循环执行一次 print，共执行 3 轮
for fruit in fruits:
    print(f"我喜欢吃{fruit}")  # f-string：在字符串中嵌入 fruit 的值

print("---")

# 遍历字符串（逐个字符）
# 字符串也是可迭代对象——for 循环会逐个取出其中的字符
for char in "Python":
    # 循环变量 char 依次被赋值为 'P', 'y', 't', 'h', 'o', 'n'
    print(f"字符: {char}")

print("---")

# 遍历字典（默认遍历键）
student = {"name": "小明", "age": 20, "score": 95}
# 直接遍历字典时，每次取出的是字典的键（key）
# 如果需要访问对应的值，需要用 student[key] 来获取
for key in student:
    print(f"{key}: {student[key]}")  # student[key] 获取键对应的值

print("---")

# 遍历字典的键值对（使用 .items()）
# .items() 方法返回 (键, 值) 的元组序列
# 使用 key, value 两个变量进行解包（unpacking），同时获取键和值
for key, value in student.items():
    print(f"{key} -> {value}")

# ---------- 模拟 for 循环的底层执行过程 ----------
print("\n=== 模拟 for 循环的底层机制 ===")
# Python 的 for 循环本质上是通过迭代器协议工作的
# 下面的代码展示了 for 循环在底层的真实执行过程
it = iter([10, 20, 30])  # 1. iter() 获取列表的迭代器对象
while True:
    try:
        val = next(it)    # 2. next() 不断获取下一个元素
        print(f"处理: {val}")  # 处理当前元素
    except StopIteration: # 3. 当没有更多元素时，迭代器抛出 StopIteration 异常
        break              # 4. 捕获异常，退出循环
print("循环结束")
# Python 的 for 循环在底层做的正是上面这几步！
# for 循环只是这个过程的语法糖——让代码更简洁易读
````

## 6. range()——"生成一个整数序列"

### 为什么要用 range？

很多时候你需要**重复 N 次**或者**生成一系列连续整数**。`range()` 就是为此而生的。

`range(start, stop, step)` 生成从 `start` 到 `stop-1`、每隔 `step` 取一个的整数序列。

### 三种用法

| 写法 | 等价序列 | 说明 |
|------|----------|------|
| `range(5)` | 0, 1, 2, 3, 4 | 从 0 到 4，共 5 个数 |
| `range(2, 6)` | 2, 3, 4, 5 | 从 2 到 5 |
| `range(1, 10, 2)` | 1, 3, 5, 7, 9 | 从 1 到 9，步长为 2 |

### 深入理解：左闭右开区间 [start, stop)

这是编程中**非常重要**的一个惯例。`range(n)` 生成 `[0, n)`——包含 0，不包含 n。

**为什么这样设计？** 有四个好处：

1. **数量直观**：`range(n)` 正好生成 n 个数——`range(10)` 就是 0~9 十个数字
2. **无缝拼接**：`range(0, 5)` + `range(5, 10)` = `range(0, 10)`，不重不漏
3. **与切片一致**：`list[0:n]` 取前 n 个 → `len(list[0:n])` 正好是 n
4. **空区间合法**：`range(0, 0)` 是合法的空区间——"不用管 0 到 0 之间的数"在数学上成立

> **质疑**："为什么不从 1 开始？显得更自然。" 因为编程中**基于 0 的索引**是普遍惯例（数组索引从 0 开始），`range(0, n)` 的结束值 n 正好对应列表的长度，这种一致性大大减少了 off-by-one 错误。

> **类比**：左闭右开就像"从第 start 个人开始排队，到第 stop 个人之前"。排到第 stop 个人时，他前面的都进去了，他自己还没进。

### 常见陷阱：range 和列表的区别

```python
r = range(5)
print(r)        # 输出：range(0, 5)，而不是 [0, 1, 2, 3, 4]
print(type(r))  # <class 'range'>，不是 list！
```

`range()` 返回的不是列表，而是一个 **range 对象**。它是惰性的——不存储所有数字，只存 `start`、`stop`、`step` 三个值。需要某个数字时，通过 `start + index * step` 当场计算。这也是为什么 `range(10**9)` 几乎不占内存（几十个字节），而 `list(range(10**9))` 会吃掉 GB 内存。

### 用 range 实现倒序遍历

```python
# 从 10 数到 1
for i in range(10, 0, -1):
    print(i)  # 10, 9, 8, ..., 1

# 注意：stop=0，不包括 0
# 如果写成 range(10, 0) 没有指定步长，默认步长是 +1，但 10 > 0，无法递增到 0 → 空序列
```

**注意**：`range(10, 0)` 是空序列！默认步长是 `+1`，但 `start=10` 已经大于 `stop=0`，所以"增不回去"，结果为空。要倒序必须显式指定 `step=-1`。

### 实用技巧

- **循环 N 次**：`for _ in range(N):` —— 用 `_` 表示不需要使用循环变量的值
- **遍历列表的索引**：`for i in range(len(lst)):` —— 但通常 `enumerate()` 更优雅
- **按步长取子序列**：`range(0, 100, 10)` 生成 0, 10, 20, ..., 90

````python
# ================================
# 2.6 range() 用法
# ================================

# ---------- 三种基本形式 ----------
# range(stop)：生成从 0 到 stop-1 的整数序列，共 stop 个数
print("range(5):", list(range(5)))           # [0, 1, 2, 3, 4]
# range(start, stop)：生成从 start 到 stop-1 的整数序列
print("range(2, 6):", list(range(2, 6)))     # [2, 3, 4, 5]
# range(start, stop, step)：从 start 开始，每次加 step，直到达到或超过 stop
# 关键：左闭右开区间 [start, stop)，不包含 stop
print("range(1, 10, 2):", list(range(1, 10, 2)))  # [1, 3, 5, 7, 9]
# step 可以是负数：从 10 开始，每次减 2，到达到或低于 stop 时停止
# 注意 stop=0 意味着生成大于 0 的最小数（即 2），但不包含 0
print("range(10, 0, -2):", list(range(10, 0, -2)))  # [10, 8, 6, 4, 2]

# ---------- 常用模式：循环 N 次 ----------
print("\n=== 循环 5 次 ===")
# 当不需要使用循环变量本身时，习惯上用 _ 或 i 作为变量名
# range(5) 生成 0, 1, 2, 3, 4 五个数，循环体执行 5 次
for i in range(5):
    # i 从 0 到 4，加 1 后显示 1 到 5，更符合人类计数习惯
    print(f"第 {i + 1} 次循环")

# ---------- 常用模式：用索引遍历列表 ----------
fruits = ["苹果", "香蕉", "橘子", "西瓜"]
print("\n=== 带索引遍历 ===")
# range(len(fruits)) 生成 0, 1, 2, 3（对应列表的索引）
# 通过 fruits[i] 可以同时拿到索引 i 和对应的元素
# 但这种方式不如 enumerate() 优雅（见下一节）
for i in range(len(fruits)):
    print(f"第 {i + 1} 个水果：{fruits[i]}")

# ---------- range 是"惰性"的 ----------
print("\nrange(10**9) 占用内存:", end=" ")
# 创建一个包含十亿个整数的 range 对象——但几乎不占用内存！
r = range(10**9)  # 10^9 = 1,000,000,000
import sys
print(f"{sys.getsizeof(r)} 字节")  # 只有几十个字节！
# 如果试图用 list(range(10**9)) 创建一个真正的列表：
# 会消耗约 8GB 内存（每个整数 8 字节 × 十亿）
print("list(range(10**9)) 会吃掉 GB 内存——千万别试！")

# 为什么 range 这么省内存？
# range 不存储所有整数，只存储 start、stop、step 三个值
# 需要某个整数时，当场计算出来：start + index * step
# 这种"需要时才计算"的策略叫"惰性求值"（lazy evaluation）
````

## 7. enumerate() 和 zip()——循环的"最佳搭档"

### enumerate() ——"既拿元素又拿序号"

上面的例子用了 `range(len(list))` 来同时拿索引和元素。但 Python 提供了一个更优雅的方式：`enumerate()`。

**它做了什么？** 把 `["a", "b", "c"]` 变成 `(0, "a"), (1, "b"), (2, "c")` 这样的**索引-元素对**。

```python
# 不优雅
for i in range(len(fruits)):
    print(i, fruits[i])

# Pythonic
for i, fruit in enumerate(fruits):
    print(i, fruit)
```

第二个写法不仅更简洁，而且直接解包了两个变量 `i` 和 `fruit`，可读性大大提升。

`enumerate()` 还可以指定起始值：`enumerate(fruits, start=1)` 从 1 开始计数——非常适合"第 1 项、第 2 项..."的场景。

### zip() ——"把两个拉链合到一起"

想象你有两条拉链带，`zip()` 就像拉链头一样，**一一对应**地把它们结合成一对一对的元组。

```python
names = ["小明", "小红", "小刚"]
scores = [95, 88, 92]
# zip(names, scores) → ("小明", 95), ("小红", 88), ("小刚", 92)
```

**重要**：如果两个列表长度不同，`zip()` 以**短的那个**为准——长出来的部分被丢弃了。Python 3.10+ 有 `zip(strict=True)` 可以在不等长时报错。

> **类比**：zip 就像给两个长度不同的木板钉钉子——短木板有多长，钉子就钉到哪，长出来的部分就空着。

### 什么时候用？

- `enumerate()`：既需要索引又需要元素时
- `zip()`：需要同步遍历多个序列时（比如学生姓名和对应的分数）
- `enumerate(zip(...))`：需要索引 + 多个序列同步遍历时

### 常见误区

**误区1：`enumerate` 返回的是一个列表吗？**

不是。`enumerate()` 返回一个 **enumerate 对象**，它是惰性的（和 range 一样）。只有当你遍历它时，才逐个生成 `(索引, 值)` 对。`list(enumerate(fruits))` 才能看到全部内容。

**误区2：`zip` 一定要解包使用**

```python
# 可以不解包，拿到的是元组
for pair in zip(names, scores):
    print(pair)  # ('小明', 95), ('小红', 88), ...

# 也可以解包成两个变量
for name, score in zip(names, scores):
    print(name, score)  # 小明 95, 小红 88, ...
```

**误区3：混淆 `enumerate` 和 `zip` 的用途**

- `enumerate(x)` = 给序列 x 的元素配上序号（一维）
- `zip(x, y)` = 把序列 x 和 y 的元素一一配对（多维）
- 两者可以组合：`enumerate(zip(x, y), start=1)`

````python
# ================================
# 2.7 enumerate() 和 zip()
# ================================

fruits = ["苹果", "香蕉", "橘子", "西瓜"]

# ---------- enumerate ----------
print("=== enumerate 基础 ===")
# enumerate() 把列表变成（索引, 元素）的配对序列
# 每次循环返回一个元组 (index, element)
# i, fruit 是对这个元组的解包赋值
for i, fruit in enumerate(fruits):
    print(f"索引 {i}: {fruit}")  # i 是自动生成的索引，fruit 是列表元素

print("\n=== enumerate(start=1) ===")
# start 参数指定索引起始值（默认从 0 开始）
# 这里从 1 开始，适合"第1个、第2个..."的人类计数场景
for i, fruit in enumerate(fruits, start=1):
    print(f"第{i}个: {fruit}")

# 如果不解包，enumerate 直接返回 (索引, 元素) 的元组
for item in enumerate(fruits):
    print(item)  # 输出：(0, '苹果'), (1, '香蕉'), ...

# ---------- zip ----------
print("\n=== zip 基础 ===")
names = ["小明", "小红", "小刚"]
scores = [95, 88, 92]
# zip() 将多个序列中位置相同的元素一一配对，组成元组
# 第一次循环：("小明", 95)，第二次：("小红", 88)，第三次：("小刚", 92)
# name, score 是对配对元组的解包
for name, score in zip(names, scores):
    print(f"{name} 考了 {score} 分")

# 三个序列一起 zip
subjects = ["语文", "数学", "英语"]
exam_results = [90, 95, 88]
# zip() 可以接受任意数量的可迭代对象
# 注意：zip 以最短的序列长度为准，超出部分被丢弃
# 这里 subjects 有 3 个元素，names 也有 3 个，exam_results 有 3 个，正好配对
for sub, name, score in zip(subjects, names, exam_results):
    print(f"{name} 的 {sub} 成绩: {score}")

# zip 解包——"反 zip"操作
print("\n=== zip 解包 ===")
pairs = [(1, 'a'), (2, 'b'), (3, 'c')]
# zip(*pairs) 相当于 zip((1, 'a'), (2, 'b'), (3, 'c'))
# * 是解包运算符，把列表中的三个元组"展开"成三个独立的参数
# 结果：所有第一个元素组成元组 (1, 2, 3)，所有第二个元素组成元组 ('a', 'b', 'c')
numbers, letters = zip(*pairs)
print("numbers:", numbers)  # (1, 2, 3)
print("letters:", letters)  # ('a', 'b', 'c')

# enumerate + zip 组合
print("\n=== enumerate + zip ===")
students = ["小明", "小红", "小刚"]
scores_list = [92, 88, 95]
# 嵌套使用：enumerate 包装 zip 的结果
# zip(students, scores_list) 生成 ("小明", 92), ("小红", 88), ("小刚", 95)
# enumerate(..., start=1) 为每组数据添加排名序号
# rank 接收序号，(name, score) 接收解包后的姓名和分数
for rank, (name, score) in enumerate(zip(students, scores_list), start=1):
    print(f"第{rank}名: {name} — {score}分")
````

## 8. 嵌套循环——"循环里面套循环"

### 为什么要嵌套？

如果一个循环解决一个维度的问题，嵌套循环解决**多个维度**的问题：
- 打印一个矩阵（行 × 列）
- 比较两个列表中的每一对元素
- **九九乘法表**——外层循环决定"行"（被乘数），内层循环决定"列"（乘数）
- 生成坐标组合 —— `(0,0), (0,1), (0,2), (1,0), ...`

> **类比**：嵌套循环就像你的日程表。外层循环是"星期一到星期五"（5 天），内层循环是"上午、下午、晚上"（3 个时段）。每天都会经历上午、下午、晚上——外层走一天，内层跑一轮。

### 执行过程

```python
for i in range(3):       # 外层：i = 0, 1, 2
    for j in range(2):   # 内层：每次外层循环，j 都从 0 到 1
        print(i, j)
```

执行顺序：
1. i=0 → 进入内层 → j=0 → print → j=1 → print → 内层结束
2. i=1 → 进入内层 → j=0 → print → j=1 → print → 内层结束
3. i=2 → 进入内层 → j=0 → print → j=1 → print → 内层结束

**总共执行外层循环 3 次 × 内层循环 2 次 = 6 次。** 当问题需要两层或多层维度时，总的迭代次数是各层循环次数的乘积。这意味着**层数越多、每层范围越大，总执行次数增长越快**——这一点在处理大数据时要特别留意。

### 九九乘法表

```
1 x 1 = 1
1 x 2 = 2   2 x 2 = 4
1 x 3 = 3   2 x 3 = 6   3 x 3 = 9
...
```

注意观察规律：
- 第 i 行有 i 个表达式
- 每行第一个数从 `1` 到 `i`，第二个数固定为 `i`

**如何从规律到代码？** 这种"每行长度不同"的嵌套循环，关键在于**内层循环的范围由外层循环变量决定**。第 i 行有 i 个表达式 → `for j in range(1, i + 1)`。

### 嵌套循环的常见陷阱

**性能意识**：嵌套循环的复杂度是 O(n×m)。如果 n 和 m 都很大（比如各 10000），总执行次数是 1 亿次——程序会明显变慢。实际编码时要思考："我真的需要两层循环吗？有没有更高效的方式？"

**break 只退出当前层**：在内层循环中使用 `break`，只会退出内层循环，外层循环不受影响。如果要想"找到目标后退出所有循环"，需要额外机制（比如标志变量或 `for-else` + `break` 的组合）。

````python
# ================================
# 2.8 嵌套循环与九九乘法表
# ================================

# ---------- 嵌套循环执行过程演示 ----------
print("=== 嵌套循环执行过程 ===")
# 外层循环：i 从 0 到 2，共 3 次
# 内层循环：每次外层循环，j 都从 0 重新开始到 1，共 2 次
# 总执行次数：3（外层）× 2（内层）= 6 次
for i in range(3):
    print(f"外层循环: i = {i}")
    # 内层循环完全包含在外层循环内部
    # 注意缩进——缩进 8 个空格表示它属于外层循环的代码块
    for j in range(2):
        print(f"  内层循环: j = {j}")
    # 这个 print 和上面的 for 对齐，属于外层循环
    # 每次外层循环结束后执行
    print(f"外层循环 {i} 结束")
    print()  # 空一行，使输出更清晰

# 执行顺序详解：
# i=0 → 进入内层 → j=0 → print → j=1 → print → 内层结束 → 打印"外层 0 结束"
# i=1 → 进入内层 → j=0 → print → j=1 → print → 内层结束 → 打印"外层 1 结束"
# i=2 → 进入内层 → j=0 → print → j=1 → print → 内层结束 → 打印"外层 2 结束"

# ---------- 九九乘法表 ----------
print("=== 九九乘法表 ===")
# i 从 1 到 9，代表乘法表的行数（被乘数）
for i in range(1, 10):       # i: 1 ~ 9（行）
    # j 从 1 到 i，代表每行的列数（乘数）
    # 关键：range(1, i+1) 使每行输出的表达式数量等于当前行号
    # 第 1 行有 1 个表达式，第 2 行有 2 个，...，第 9 行有 9 个
    for j in range(1, i + 1): # j: 1 ~ i（列，每行 i 个）
        # {j}x{i}={i*j:2d}：输出格式为"乘数x被乘数=乘积"
        # :2d 表示整数占 2 个字符宽度（右对齐），使输出对齐美观
        # end="  " 用两个空格代替默认的换行符，使同一行的表达式排在一行
        print(f"{j}x{i}={i * j:2d}", end="  ")
    # 内层循环结束后执行 print()——默认输出换行符
    # 作用：结束当前行，下一行的表达式从新行开始
    print()  # 换行

# ---------- 不换行输出说明 ----------
# print() 默认在末尾加换行符 \n，所以每次 print 都换到下一行
# end="  " 告诉 print 用两个空格代替换行
# 这样同一行的多个表达式就在同一行显示了
# 内层循环结束后，print() 换行

# ---------- 倒序九九乘法表 ----------
print("\n=== 倒序九九乘法表 ===")
# range(9, 0, -1)：从 9 开始，每次减 1，到 0 之前停止
# 所以 i 依次为 9, 8, 7, ..., 1（从大到小）
for i in range(9, 0, -1):
    # j 从 1 到 i——逐行减少
    for j in range(1, i + 1):
        print(f"{j}x{i}={i * j:2d}", end="  ")
    print()  # 每行结束后换行
````

## 9. while 循环——"直到条件不再满足"

### 什么时候用 for，什么时候用 while？

**简单判断标准**：
- **知道要循环多少次** → 用 `for`（比如遍历列表、range）
- **不知道要循环多少次，但有"继续的条件"** → 用 `while`（比如"直到用户输入正确"）

> **类比**：for 循环像"按菜单做菜"——你知道有几道菜，做完为止。while 循环像"给锅里的水加热"——你不知道多久会沸腾，但你知道"只要没沸腾就继续加热"。

### while 的结构

```python
while 条件:
    循环体
```

执行过程：
1. 判断条件 → True → 执行循环体 → 回到第 1 步
2. 判断条件 → False → 退出循环

### 无限循环——程序员又爱又恨的东西

```python
while True:        # 条件永远为真
    print("停不下来！")
```

**无限循环不一定是 bug！** 很多程序**需要**永远运行：
- 游戏的主循环："反复做：读取输入 → 更新画面 → 渲染"
- Web 服务器："反复做：等待请求 → 处理请求 → 返回响应"
- 监控程序："反复做：检查温度 → 如果过高就报警"

但这些无限循环通常有一个**退出条件**，通过 `break` 跳出。

**真正的陷阱是"非刻意的无限循环"**——你想让循环结束，但忘记更新条件变量，导致条件永远为真。比如：

```python
i = 0
while i < 10:
    print(i)
    # 忘了 i += 1 → i 永远是 0 → 0 < 10 永远为 True → 无限循环！
```

> **防止方法**：先写 `while` 的框架，紧接着写条件变量的更新语句，再写循环体的其他内容。

### while 的常见坑

1. **忘记更新条件变量**——条件永远为真，变成无限循环
2. **条件一开始就是 False**——循环体一次都不执行
   ```python
   while False:
       print("永远不会执行")
   ```
   这不是 bug，有时是有意为之（虽然很少见）。
3. **条件永远无法满足**——比如 `while x != 10`，但你**减少** x 的值，而 x 初始值小于 10，x 永远到不了 10，循环永不结束。

### while 的"一步初始化"模式

用 while 循环时，经常需要在循环开始前**初始化条件变量**，然后在循环体内**更新条件变量**：

```python
# 三步走：
# 1. 初始化条件变量
count = 0
# 2. while 条件
while count < 5:
    print(count)
    # 3. 更新条件变量（绝对不能忘！）
    count += 1
```

而 for 循环把这三步合成了一个语句：`for count in range(5)`。所以**如果确实知道循环次数，for 循环更安全**——你不可能"忘记更新"循环变量。

````python
# ================================
# 2.9 while 循环
# ================================

# ---------- 基本 while ----------
print("=== while 基础 ===")
count = 0  # 初始化循环变量（条件变量）
# while 循环：只要条件 count < 5 为 True，就继续执行循环体
while count < 5:
    print(f"计数: {count}")
    count += 1  # 一定要更新条件变量！否则 count 永远是 0，导致无限循环
# 执行过程：count=0 → 0<5 True → print → count=1 → 1<5 True → ...
# → count=4 → 4<5 True → print → count=5 → 5<5 False → 退出循环

# ---------- 对比 for 和 while ----------
# 用 for 实现"打印 0~4"
print("\n=== for 实现 ===")
# for 循环：当知道确切循环次数时更合适
for i in range(5):
    print(i)

# 用 while 实现相同功能
print("\n=== while 实现 ===")
i = 0  # 需要手动初始化循环变量
while i < 5:
    print(i)
    i += 1  # 需要手动更新循环变量
# 对比：for 更简洁且不容易出错（不需要手动管理变量）

# ---------- while 更适合的场景：不确定循环次数 ----------
import random

print("\n=== 猜硬币（while 用法示范） ===")
# while 适合"不知道循环多少次，但有继续条件"的场景
target = random.randint(1, 100)  # 随机生成 1~100 的目标值
guess = -1  # 初始化为一个不可能的值，确保进入循环
attempts = 0
# 条件：当猜测值不等于目标值时继续循环
# 因为我们不知道要猜多少次才中，所以用 while 比 for 更合适
while guess != target:
    guess = random.randint(1, 100)  # 模拟随机猜测
    attempts += 1  # 每猜一次计数加 1
print(f"猜了 {attempts} 次才猜中 {target}！")

# ---------- 无限循环的实用例子 ----------
# 很多实际程序需要"永远运行"：Web 服务器、游戏主循环、监控程序
# 使用 while True + break 退出是常见模式
print("\n=== 简易菜单系统 ===")
# while True 创建无限循环——条件永远为真
# 需要通过内部的 break 语句来退出
while True:
    print("\n1. 查看余额")
    print("2. 存款")
    print("3. 取款")
    print("0. 退出")
    # input() 从控制台读取用户输入（返回字符串）
    choice = input("请选择操作：")
    if choice == "0":
        print("再见！")
        break  # break 退出当前所在的 while 循环
    elif choice == "1":
        print("您的余额是 100 元（演示）")
    elif choice == "2":
        print("存款功能（演示）")
    elif choice == "3":
        print("取款功能（演示）")
    else:
        # 如果用户输入的不是 0-3，提示重新输入
        # 注意：这里没有 break，循环会继续执行
        print("无效选择，请重试")
````

## 10. break 和 continue——循环的"遥控器"

### break ——"立即停止，马上退出"

**类比**：你在电影院里看电影。电影放到一半时，你发现这是你一个月前看过的片子。你站起身就走了——**不再继续看，也不再回来**。这就是 `break`。

`break` 会让 Python **立即退出当前所在的循环**，不再执行剩下的循环体，也不再继续下一次循环。

**注意**：`break` **只退出它所在的那一层循环**。如果在嵌套循环内部用 `break`，只有内层循环被退出。

> **另一个类比**：break 就像在图书馆找一本书。你从第一个书架开始找，找到目标就 stop——你不会继续找完剩下的书架。

### continue ——"跳过这次，继续下一次"

**类比**：你在课堂上点名。喊到"张三"时，张三请假了。你就在名单上跳过他，继续喊下一个名字。这就是 `continue`。

`continue` 会让 Python **立即结束本轮循环**，跳过后面所有的代码，直接进入下一次循环。

### 对比

| | break | continue |
|---|---|---|
| 效果 | 完全退出循环 | 跳过本轮，继续下一轮 |
| 类比 | 退场不看了 | 跳过这个，继续下一个 |
| 对后续影响 | 循环不再执行 | 循环继续执行 |
| 适用场景 | 找到了想要的东西 | 当前值不符合要求 |

### 常见模式

```python
# break：找到第一个满足条件的就停
for student in students:
    if student.id == target_id:
        print(f"找到了：{student.name}")
        break  # 找到了就不用再找了

# continue：跳过不需要的
for num in range(100):
    if num % 2 == 0:
        continue  # 偶数跳过
    print(num)    # 只处理奇数
```

### 常见陷阱

**1. break vs continue 混淆**

初学者最常见的错误之一：在应该用 `break` 的地方用了 `continue`，或者在应该用 `continue` 的地方用了 `break`。

**记住**：
- `break` = "我完事了，整个循环结束"
- `continue` = "这个我不要了，循环接着跑"

**2. break 在嵌套循环中只退出当前层**

```python
for i in range(3):
    for j in range(3):
        if j == 1:
            break  # 只退出 for j 循环！
    # 程序继续执行到这里——外层 for i 不受影响
```

如果你想要"找到目标就退出所有循环"，有两种方案：
- **方案 A**：使用标志变量（flag variable）
- **方案 B**：把循环封装到函数中，用 `return` 退出

**3. continue 在 while 循环中要小心**

在 while 循环中使用 `continue` 时，如果条件变量的更新在 `continue` 之后，会导致条件变量不被更新，造成无限循环：

```python
i = 0
while i < 10:
    if i % 2 == 0:
        continue  # 跳过了 i += 1！i 永远为 0 → 无限循环！
    print(i)
    i += 1  # 被 continue 跳过了！
```

**解决方案**：在 `continue` 之前更新条件变量，或者用 `for` 循环替代（更安全）。

````python
# ================================
# 2.10 break 和 continue
# ================================

# ---------- break 演示 ----------
print("=== break：找到 3 就停 ===")
# break：立即退出当前循环，不再执行剩余迭代
for num in range(1, 10):
    if num == 3:
        print("找到 3！停止循环")
        break  # break 被执行，循环立即结束
    # 注意：当 num=3 时，下面的 print 不会被执行
    # break 跳过了当前循环中 break 之后的所有代码
    print(f"检查: {num}")
# break 之后程序继续执行循环后面的代码
print("循环结束")
# 输出：检查 1 → 检查 2 → 找到 3 → 循环结束
# num=4 到 9 的迭代都被跳过了

print()

# ---------- continue 演示 ----------
print("=== continue：跳过偶数 ===")
# continue：跳过本次循环的剩余代码，直接进入下一次迭代
for num in range(1, 11):
    # % 是取模运算符（求余数）
    # num % 2 == 0 表示 num 是偶数
    if num % 2 == 0:
        continue  # 跳过本次循环后面的代码，进入下一轮
    # 只有当 continue 没有被执行时才会到这里
    # 相当于只处理奇数，不处理偶数
    print(f"奇数: {num}")

print()

# ---------- break 只退出当前层循环 ----------
print("=== break 在嵌套循环中 ===")
# 重要概念：break 只退出它所在的那一层循环
# 外层循环不受内层 break 的影响
for i in range(3):
    print(f"外层 i={i}")
    for j in range(3):
        if j == 1:
            print("  内层遇到 j=1，break")
            break  # 只退出内层 for j 循环！外层 for i 继续运行
        print(f"  内层 j={j}")
    # 内层 break 后程序会跳到这里——外层循环继续执行
    print(f"外层循环 {i} 继续执行")
print("所有循环结束")

print()

# ---------- 实用案例：检查素数 ----------
num = 17
is_prime = True  # 先假设是素数，找到因子后改为 False
# 判断素数：检查 2 到 sqrt(num) 之间是否有因子
# 如果 num 有因子，必定有一个因子 <= sqrt(num)
# 这样可以减少检查次数，提高效率
for i in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
        # 找到一个因子，说明不是素数
        print(f"{num} 能被 {i} 整除，不是素数")
        is_prime = False
        break  # 找到一个因子就够了！不需要继续检查——提高效率
# 如果没有被 break（没有找到因子），is_prime 仍为 True
if is_prime:
    print(f"{num} 是素数")

# ---------- continue 实用：数据清洗 ----------
data = [1, 0, -5, 3, 0, 7, -2, 0, 9]
print("\n=== 用 continue 过滤零值 ===")
valid = []  # 创建一个空列表，用于存储过滤后的数据
for val in data:
    if val == 0:
        continue  # 遇到 0 就跳过，不加入 valid 列表
    valid.append(val)  # 只有非零值才会执行到这里
print(f"原始数据: {data}")
print(f"去掉零值: {valid}")
````

## 11. Python 的独门绝技：for-else 和 while-else

### 这是 Python 独有的特性

绝大多数编程语言没有这个特性。即使是 Python 老手，也有很多人不知道或不理解它。

**基本语法：**

```python
for 变量 in 序列:
    循环体
else:
    循环正常结束时执行
```

**关键问题是：** 什么叫"正常结束"？

- **正常结束**：循环因为迭代完了所有元素（for）或条件变为 False（while）而退出
- **非正常结束**：循环因为 `break` 而退出

### 可以这样理解

`else` 子句相当于说："如果循环**没有被 break 打断**，最后来做这件事。"

就像一个巡检员：
- 走完一整条生产线，检查所有产品 → 没发现问题 → 报告"一切正常"（执行 else）
- 中途发现一个次品 → break 打断巡检 → **不**报告"一切正常"（不执行 else）

### 为什么需要这个特性？

在没有 for-else 的情况下，如果要实现"检查所有元素，如果没找到问题就做 X"，你需要一个标志变量：

```python
found = False
for item in items:
    if condition(item):
        found = True
        break
if not found:
    do_something()
```

用 for-else，代码更清晰：

```python
for item in items:
    if condition(item):
        break
else:
    do_something()  # 只有没被 break 时才会执行
```

### 实际案例：搜索

```python
# 如果找到了目标，break 退出 → else 不执行
# 如果没找到目标，循环自然结束 → else 执行
for student in students:
    if student.id == target:
        print("找到了")
        break
else:
    print("没有这个学生")  # 只有循环没有被 break 才会执行
```

> **记忆技巧**：`else` 中的代码是"没被抓到现行"时的奖励——只有成功逃跑（没被 break 打断）才能拿到。

### 常见陷阱

**误解1**：以为 else 在循环**总是**执行 —— 不对！只有不 break 时才执行。

**误解2**：以为 else 只和 `if` 配对 —— Python 中 `else` 还可以和 `for`、`while`、`try` 配对，各有不同的含义。

**误解3**：把 `else` 和 `finally` 搞混 —— `finally` 是无论是否 break **都会**执行的（用在 `try` 语句中），而 `else` 是只有不 break 才执行的。

### 实用场景：素数判断

```python
for n in range(2, 100):
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            break  # 有因子，不是素数
    else:
        print(n, "是素数")  # 没有 break 执行到这里
```

这里的 `else` 和 `for` 配对，含义是："如果内层循环没有被 `break`（即没找到因子），就执行 else"。这是 for-else 在实际中最经典的用例之一。

````python
# ================================
# 2.11 for-else 和 while-else
# ================================

# ---------- for-else：搜索示例 ----------
print("=== for-else：在名单中搜索 ===")
students = ["小明", "小红", "小刚", "小丽"]

# 模拟用户输入——在实际运行时取消注释下面这行来手动输入
# search = input("请输入要查找的学生姓名：")
# 为了演示效果，先使用一个固定的值：
search = "小刚"  # 运行时可改成其他名字来测试不同路径

# for-else 结构：else 在 for 循环正常结束后执行
# "正常结束" = 循环不是因为 break 而结束
for s in students:
    if s == search:
        print(f"找到了：{search}")
        break  # 找到了 → break 退出 → else 块不会执行
else:
    # 只有当循环自然结束（没有被 break）时才执行
    # 自然结束 = 遍历完了 students 中的所有元素都没找到
    print(f"没有找到：{search}")

print()

# ---------- while-else 示例 ----------
print("=== while-else ===")
count = 0
while count < 3:
    print(f"count = {count}")
    count += 1
    # 取消注释下面的代码来测试 break 的效果：
    # if count == 1:
    #     print("被 break 了！")
    #     break  # 如果 break 了，else 不会执行
else:
    # 只有在 while 条件变为 False 而退出时才执行
    # 如果是被 break 退出的，不执行
    print("循环正常结束，没有被打断")

print()

# ---------- 真实案例：检查所有数字是否为正 ----------
print("=== 检查列表是否全为正数 ===")
def check_all_positive(numbers):
    for n in numbers:
        if n <= 0:
            # 发现非正数，立即报告并 break
            # 注意：只要有一个非正数，就可以下结论了，不需要继续检查
            print(f"发现非正数: {n}")
            break  # break 退出 → else 不执行
    else:
        # 只有在循环遍历完所有元素（即所有数都 > 0）时才执行
        # 如果中途 break 了，不会执行这里的 else
        print("所有数字都是正数！")

check_all_positive([3, 5, 7, 9])    # 全为正数 → 遍历完 → 执行 else
check_all_positive([3, -1, 7, 9])   # 遇到 -1 → break → 不执行 else
check_all_positive([0, 5, 7])       # 遇到 0 → break → 不执行 else
````

## 12. 🎯 综合练习：猜数字游戏

现在我们把今天学的所有知识点串起来，写一个完整的猜数字游戏。

### 游戏规则

1. 程序随机生成一个 1~100 之间的整数
2. 玩家输入猜测的数字
3. 程序告诉玩家猜大了还是猜小了
4. 玩家继续猜，直到猜中为止
5. 游戏结束时显示总共猜了多少次
6. **额外体验**：猜错时用 `continue` 提示再猜，猜中时用 `break` 退出循环

### 涉及的知识点

| 知识点 | 用在哪儿 |
|--------|----------|
| `while True` | 无限循环，直到猜中 |
| `if/elif/else` | 判断猜大、猜小、猜中 |
| `break` | 猜中时退出循环 |
| `input()` + `int()` | 接收并转换用户输入 |
| `f-string` | 格式化输出结果 |
| `random.randint()` | 生成随机数 |

### 先思考，再运行

在看代码之前，先想清楚下面的伪代码你要怎么写：

```
1. 生成随机数 target
2. 初始化 attempts = 0
3. 无限循环：
   a. 接收用户输入
   b. attempts += 1
   c. 如果 猜中 → 恭喜 → break
   d. 如果 猜大 → 提示"大了"
   e. 如果 猜小 → 提示"小了"
4. 显示总次数
```

### 代码中涉及的新概念

**`try-except` 异常处理**：在代码中你可能看到了 `try` 和 `except`——这是 Python 的异常处理机制。当 `int(input(...))` 尝试把用户输入的字符串转换成整数时，如果用户输入了"abc"之类的非数字内容，`int()` 会抛出 `ValueError`。`try-except` 捕获这个错误并提示用户重新输入，程序不会崩溃。

这是 Python 中**推荐的做法**——与其检查用户输入是否合法（"事前检查"），不如尝试执行并捕获可能的错误（"请求原谅比请求许可更容易"——EAFP 原则）。

### 常见陷阱（在写猜数字游戏时要注意）

1. **死循环**：如果忘了在猜中时 `break`，程序会无限循环——即使猜中了也不会退出
2. **类型转换错误**：`input()` 返回的是字符串，直接和数字比较会导致 TypeError 或意外的结果（在 Python 3 中会直接报错）
3. **逻辑顺序**：先判断 `guess > target` 还是先判断 `guess < target` 不影响结果，但要确保猜中对的情况被正确处理
4. **计数逻辑**：`attempts` 应该在每次输入有效数字后递增，而不是每次循环都递增（无效输入不应该算一次猜测）

### ⭐ 挑战任务（学有余力的同学试试）

1. 添加难度选择：简单（1-50）/ 普通（1-100）/ 困难（1-200）
2. 限制猜测次数，超过次数游戏失败（提示：用另一个变量跟踪剩余次数）
3. 记录玩家的历史猜测并显示（提示：用一个列表存储每次的猜测值）
4. 猜完后问"是否再玩一局？"，输入 y 重新开始（提示：用外层 while 循环包裹整个游戏）

````python
# ================================
# 2.12 猜数字游戏
# ================================

# 导入 random 模块，用于生成随机数
import random

print("=" * 40)
print("         🎯 猜数字游戏")
print("=" * 40)
print("我已经想好了一个 1~100 之间的数字。")
print("你有无限次机会，来猜猜看吧！\n")

# 生成随机目标数字：randint(1, 100) 返回 1 到 100 之间的随机整数（包含两端）
target = random.randint(1, 100)
attempts = 0  # 初始化猜测次数计数器

# while True 创建无限循环——适合不确定循环次数的场景
# 循环通过内部的 break 语句退出（猜中时）
while True:
    # 获取用户输入：input() 返回字符串，需要用 int() 转换为整数
    # 如果用户输入的不是有效整数，int() 会抛出 ValueError 异常
    # 使用 try-except 来捕获并处理这个异常
    try:
        guess = int(input("请输入你的猜测（1~100）："))
    except ValueError:
        # 当 int() 转换失败时执行这里（比如用户输入了"abc"）
        print("请输入有效的整数！")
        continue  # 跳过本次循环剩余代码，重新要求输入

    attempts += 1  # 每成功获取一次有效猜测，计数加 1

    # 核心逻辑：判断猜大、猜小、还是猜中
    if guess > target:
        # 玩家猜的数字比目标大：给出提示
        print("▼ 猜大了，再小一点！")
    elif guess < target:
        # 玩家猜的数字比目标小：给出提示
        print("▲ 猜小了，再大一点！")
    else:
        # guess == target：猜中了！
        print(f"\n🎉 恭喜你猜中了！就是 {target}！")
        print(f"你总共猜了 {attempts} 次。")
        break  # 猜中了，用 break 退出 while 循环

print("\n游戏结束，感谢游玩！")

# ⭐ 挑战任务（学有余力的同学试试）：
# 1. 添加难度选择：简单(1-50) / 普通(1-100) / 困难(1-200)
# 2. 限制猜测次数，超过次数游戏失败
# 3. 记录玩家的历史猜测并显示
# 4. 猜完后问"是否再玩一局？"，输入 y 重新开始
````
