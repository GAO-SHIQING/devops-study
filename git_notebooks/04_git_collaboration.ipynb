## Day 22:远程协作核心

### fetch vs pull

```
git fetch origin    → 下载远程数据到 .git/refs/remotes/origin/
                      (只下载,不合并,安全)

git pull            → git fetch + git merge
                      或 git fetch + git rebase (--rebase)
                      (下载并合并,一步到位)
```

**推荐:先 fetch,看情况再决定 merge 还是 rebase。**

```bash
git config --global pull.ff only  # 只允许 fast-forward pull
```

## Day 23:Pull Request 完整流程

1. 创建功能分支:`git checkout -b feature/user-login`
2. 开发并提交:多次 `git add + git commit`(小步提交)
3. rebase 整理:`git rebase -i HEAD~5`(清理零碎 commit)
4. 同步主分支:`git rebase main`(或 merge)
5. 推送分支:`git push -u origin feature/user-login`
6. 创建 PR:在 GitHub/GitLab 上发起 Pull Request
7. Code Review:reviewer 提意见,修改后 push
8. CI 通过:等待 CI 跑完所有检查
9. 合并:选择 merge / squash merge / rebase merge
10. 清理:`git branch -d feature/user-login`(删除本地分支)

## Day 24:三种合并策略对比

| 策略 | 效果 | 历史 | 适用 |
|------|------|------|------|
| `git merge` | 保留所有 commit + merge commit | 分叉 → 合并 | 需要追溯完整历史 |
| `git merge --squash` | 所有变更压成一个 commit | 线性 | feature 的 commit 太零碎 |
| `git rebase + merge` | 线性历史,无 merge commit | 完全线性 | 追求干净历史的小团队 |

## Day 25:分支策略

| 策略 | 分支模型 | 适用 |
|------|---------|------|
| **GitHub Flow** | main + feature 分支 | 持续部署、SaaS |
| **Git Flow** | main+develop+feature+release+hotfix | 有明确发布周期 |
| **GitLab Flow** | main + 环境分支 | 多环境部署 |
| **Trunk-Based** | 只往 main 提交 + 短生命周期分支 | 成熟 DevOps 团队 |

## Day 26:提交信息规范

### Conventional Commits

```
<type>(<scope>): <description>
```

类型:feat / fix / docs / refactor / test / chore / perf / ci

**好的提交信息:**
- `feat(login): add OAuth2 support for Google login`
- `fix(db): resolve connection timeout under high load`

**不好的提交信息:**
- `fixed bug`
- `update code`
- `wip`

## Day 27:Git 配置优化

推荐的 ~/.gitconfig 配置:

```bash
git config --global pull.ff only
git config --global merge.conflictstyle diff3
git config --global rebase.autosquash true
git config --global rerere.enabled true
git config --global diff.colorMoved zebra
git config --global fetch.prune true
git config --global branch.sort -committerdate
```

推荐 alias:

```bash
git config --global alias.lg 'log --graph --oneline --decorate'
git config --global alias.unstage 'restore --staged .'
git config --global alias.undo 'reset --soft HEAD~1'
git config --global alias.amend 'commit --amend --no-edit'
```

## Day 28:第4周综合练习

### 团队 Git 规范文档模板

````python
print("""
# 团队 Git 工作规范

## 一、分支命名

- feature/<name>     新功能: feature/oauth-login
- fix/<issue-id>      修 bug: fix/issue-42
- hotfix/<desc>       紧急修复: hotfix/critical-leak
- release/<version>   发布分支: release/v2.1.0

## 二、提交信息格式

<type>(<scope>): <subject>

类型: feat, fix, docs, refactor, test, chore, perf, ci

## 三、PR 流程

1. 从 main 创建分支
2. 开发,小步提交
3. 推送前用 git rebase -i 整理
4. 推送并创建 PR
5. 至少 1 人 approve
6. CI 全部通过
7. 使用 squash merge 合并到 main

## 四、禁止事项

- 禁止直接 push 到 main
- 禁止 force push 到共享分支
- 禁止提交密码/密钥/Token
- 禁止一个大 PR 包含 10+ 不相关的改动

## 五、紧急 hotfix 流程

1. 从 main 创建 hotfix 分支
2. 修复 → push → PR → merge
3. 打 tag
4. cherry-pick 修复到其他活跃发布分支
5. 事后写 postmortem
""")

print("=" * 60)
print("Git 第4周核心收获:")
print("1. fetch 安全;pull 自动合并可能产生意外")
print("2. PR 是代码质量的守门员,不是走过场")
print("3. 三种合并策略各有适用场景,团队应统一规范")
print("4. 分支策略选择取决于团队规模和部署节奏")
print("5. 规范不是约束,是让团队协作更高效的基础设施")
print()
print("Git 深入 4 周学习完成!")
print("=" * 60)
````
